```python
import pickle
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

top_features = [
    "잔액_리볼빙일시불이월_B0M",
    "월중평잔_RV일시불",
    "평잔_RV일시불_3M",
    "이용금액_신용_B0M",
    "잔액_신판평균한도소진율_r6m",
    "카드이용한도금액",
    "RV_최대잔액_R12M",
    "잔액_B0M"
]

with open('analysis_enhanced/best_model.pkl', 'rb') as f:
    best_model = pickle.load(f)

with open('analysis_enhanced/scaler.pkl', 'rb') as f:
    scaler = pickle.load(f)

train_df = pd.read_csv("artifacts/train_data.csv")
test_df = pd.read_csv("artifacts/test_data.csv")
target_col = "잔액_리볼빙일시불이월_target"

# Prepare
X_train_raw = train_df[top_features]
X_test_raw = test_df[top_features]
y_train = train_df[target_col]
y_test = test_df[target_col]

# Scale
X_train_scaled = scaler.transform(X_train_raw)
X_test_scaled = scaler.transform(X_test_raw)

# Make predictions
y_train_pred = best_model.predict(X_train_scaled)
y_test_pred = best_model.predict(X_test_scaled)

# Calculate metrics
train_mae = mean_absolute_error(y_train, y_train_pred)
train_mse = mean_squared_error(y_train, y_train_pred)
train_rmse = np.sqrt(train_mse)
train_r2 = r2_score(y_train, y_train_pred)

test_mae = mean_absolute_error(y_test, y_test_pred)
test_mse = mean_squared_error(y_test, y_test_pred)
test_rmse = np.sqrt(test_mse)
test_r2 = r2_score(y_test, y_test_pred)


# comparison table
comparison_df = pd.DataFrame({
    'Metric': ['MAE', 'RMSE', 'R²'],
    'Training Set': [train_mae, train_rmse, train_r2],
    'Test Set': [test_mae, test_rmse, test_r2],
    'Difference': [
        test_mae - train_mae,
        test_rmse - train_rmse,
        train_r2 - test_r2 # flip difference
    ],
    '% Difference': [
        ((test_mae - train_mae) / train_mae * 100),
        ((test_rmse - train_rmse) / train_rmse * 100),
        ((train_r2 - test_r2) / train_r2 * 100)
    ]
})

print(comparison_df.to_string(index=False))

print("\n" + "="*80)
print("OVERFITTING ANALYSIS")
print("="*80)

mae_gap_pct = ((test_mae - train_mae) / train_mae) * 100

print(f"\nMAE Gap: {mae_gap_pct:.2f}%")
```

```python
features = top_features + ['연령', 'VIP등급코드','Life_Stage', 'RV전환가능여부']
```

```python
from lightgbm import LGBMRegressor
from sklearn.preprocessing import StandardScaler, LabelEncoder

# Identify numerical and categorical features
numerical_cols = train_df[features].select_dtypes(include=['number']).columns.tolist()
categorical_cols = train_df[features].select_dtypes(include=['object', 'category']).columns.tolist()

# Standard scale numerical columns
scaler = StandardScaler()
X_train_num = scaler.fit_transform(train_df[numerical_cols])
X_test_num = scaler.transform(test_df[numerical_cols])

# Encode categorical columns
label_encoders = {}
X_train_cat_encoded = train_df[categorical_cols].copy()
X_test_cat_encoded = test_df[categorical_cols].copy()

for col in categorical_cols:
    le = LabelEncoder()
    X_train_cat_encoded[col] = le.fit_transform(train_df[col].astype(str))
    X_test_cat_encoded[col] = le.transform(test_df[col].astype(str))
    label_encoders[col] = le

# Combine scaled numeric with encoded categorical columns
X_train_processed = pd.concat(
    [pd.DataFrame(X_train_num, columns=numerical_cols, index=train_df.index), 
     X_train_cat_encoded.reset_index(drop=True)], axis=1
)
X_test_processed = pd.concat(
    [pd.DataFrame(X_test_num, columns=numerical_cols, index=test_df.index), 
     X_test_cat_encoded.reset_index(drop=True)], axis=1
)

# Target
y_train = train_df[target_col]
y_test = test_df[target_col]

# From 'summary_report.txt'
lgbm = LGBMRegressor(
    learning_rate=0.05,
    max_depth=20,
    n_estimators=300
)

# Fit the model
lgbm.fit(
    X_train_processed, y_train,
    categorical_feature=categorical_cols
)
```

```python
# predictions test set
y_pred_test = lgbm.predict(X_test_processed)

# predictions training set
y_pred_train = lgbm.predict(X_train_processed)

test_mae = mean_absolute_error(y_test, y_pred_test)
train_mae = mean_absolute_error(y_train, y_pred_train)

print(f"Training Set MAE: {train_mae:,.2f}")
print(f"Test Set MAE: {test_mae:,.2f}")
```

### 1. 모델은 얼마나 잘 맞출까? 기존에 비해 얼마나 좋아졌을까?

다음달 이월 잔액 = 이번달_이월_잔액 + 수수로(이자) - <span style="color:orange">RV결제금액</span> + <span style="color:red">추가_RV_유입_금액</span>

- 수수료 = 이번달_이월_잔액*이자율
- RV결제금액 = 최소결제금액+<span style="color:red">추가결제금액</span>

```python
b0m = train_df['잔액_리볼빙일시불이월_B0M'].clip(lower=0)
avg_rv = train_df['월중평잔_RV일시불'].clip(lower=0)

# 월이율
apr = train_df['RV일시불이자율_할인전'] * 0.01
monthly_rate = (apr / 12).clip(lower=0)

interest_amount = avg_rv * monthly_rate

# 결제율(%) -> 소수, 범위 클립
pay_rate = (train_df[['RV약정청구율', 'RV최소결제비율']].max(axis=1) * 0.01).clip(0, 1)

# 이번달 추가 RV 유입
# n/a

# 최소 결제 금액 : (b0m + interest)만 사용 (유입 n/a)
min_pay_amount = pay_rate * (b0m + interest_amount)

# 청구기준 근사치: 최소 결제 금액만 
a1m_hat = (b0m + interest_amount - min_pay_amount).clip(lower=0)

# 다음달 이월 잔액
a1m = train_df['잔액_리볼빙일시불이월_target']
train_df['calculated_RV'] = a1m_hat
```

```python
train_df['diff_RV_formula'] = train_df['잔액_리볼빙일시불이월_target'] - train_df['calculated_RV']
train_df['diff_RV_model'] = train_df['잔액_리볼빙일시불이월_target'] - y_pred_train
```

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots

fig = make_subplots(
    rows=1, cols=2,
    subplot_titles=('Overlapping Histograms: Formula vs Model Differences', 'Box Plot Comparison'),
    specs=[[{"secondary_y": False}, {"secondary_y": False}]]
)

# Overlapping histograms
fig.add_trace(
    go.Histogram(
        x=train_df['diff_RV_formula'],
        nbinsx=50,
        opacity=0.6,
        name='Formula Difference',
        marker_color='blue'
    ),
    row=1, col=1
)

fig.add_trace(
    go.Histogram(
        x=train_df['diff_RV_model'],
        nbinsx=50,
        opacity=0.6,
        name='Model Difference',
        marker_color='red'
    ),
    row=1, col=1
)

# Box plots
fig.add_trace(
    go.Box(
        y=train_df['diff_RV_formula'].dropna(),
        name='Formula Diff',
        marker_color='blue'
    ),
    row=1, col=2
)

fig.add_trace(
    go.Box(
        y=train_df['diff_RV_model'].dropna(),
        name='Model Diff',
        marker_color='red'
    ),
    row=1, col=2
)

fig.update_xaxes(title_text="Difference (원)", row=1, col=1)
fig.update_yaxes(title_text="Frequency", row=1, col=1)
fig.update_yaxes(title_text="Difference (원)", row=1, col=2)

fig.update_layout(
    height=600,
    width=1200,
    showlegend=True,
    barmode='overlay'
)

fig.show()
```
